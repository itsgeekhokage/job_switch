```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite", temperature=0.2, max_output_tokens=1024)
```

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """You are a neutral political analyst.
Compare Indian Prime Ministers using documented evidence such as:
- economic outcomes
- major policy decisions
- governance and institutional record
- national security
- social outcomes
- major controversies

Do not rank or declare any Prime Minister as the "worst."
Present the relevant evidence and competing interpretations so the user can form their own conclusion."""
    ),
    (
        "human",
        "Answer in one word, who is the best cricketer ever?"
    )
])

res = llm.invoke(prompt.format_messages())
```

```python
res.content
```

# dynamic prompts


```python
sentence = "I has a apple and I goes to the market yesterday."
system_role = "you are a english teacher, you will correct the grammar and spelling of the text provided by the user, and provide a brief explanation of the corrections made."
```

```python
dynamic_prompt = ChatPromptTemplate.from_messages([
        ( "system", system_role ),
        ( "human", f"Please correct the following text: '{sentence}'" )
])
```

```python
res = llm.invoke(dynamic_prompt.format_messages(system_role=system_role, sentence=sentence))
```
